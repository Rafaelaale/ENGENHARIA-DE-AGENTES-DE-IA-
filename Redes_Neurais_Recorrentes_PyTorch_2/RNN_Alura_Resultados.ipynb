# RNN Alura - Treinamento validado

Este notebook documenta os resultados da execucao real realizada pelo PowerShell.

O notebook original corrigido esta salvo separadamente.


```python
import json
from pathlib import Path
resultados = json.loads(Path('RNN_Alura_Validado/resultados.json').read_text(encoding='utf-8'))
print('Categorias:', len(resultados['categorias']))
print('Acuracia de teste:', resultados['acuracia_teste'])
```

```python
import matplotlib.pyplot as plt
historico = resultados['historico']
plt.plot([x['epoca'] for x in historico], [x['perda'] for x in historico], marker='o')
plt.xlabel('Epoca')
plt.ylabel('Perda media')
plt.title('Treinamento RNN')
plt.grid(True)
plt.show()
```

```python
for nome, previsoes in resultados['previsoes'].items():
    print('\n', nome)
    for item in previsoes:
        print(item['categoria'], f"{item['probabilidade']:.2%}")
```
